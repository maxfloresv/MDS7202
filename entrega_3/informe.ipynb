### 📌 Análisis individual [3.0 puntos]

````python
import pandas as pd
from sklearn.metrics import (
  accuracy_score, 
  precision_score, 
  recall_score, 
  f1_score
)

DATA_DIR = './data'
PREDS_DIR = './predictions'

clients = pd.read_parquet(f'{DATA_DIR}/clientes.parquet')
products = pd.read_parquet(f'{DATA_DIR}/productos.parquet')
# Even if some clients and products were eliminated from the train dataset, they
# are good (or bad) predictions too, because they were part of the original data.
universe_size = len(clients) * len(products)

def get_batch_performance(batch_id: int) -> dict:
  """
  Gets the performance for a specific batch.

  Parameters
  ----------
  batch_id : int
    The batch to evaluate.

  Returns
  -------
  dict
    The metrics for the evaluated batch.
  """
  data = pd.read_parquet(f"{DATA_DIR}/batch_t{batch_id}.parquet")
  data = data.groupby(['customer_id', 'product_id']).size().reset_index()
  true_pairs = set(map(
    tuple, 
    data[['customer_id', 'product_id']].values
  ))

  preds = pd.read_csv(
    f'{PREDS_DIR}/preds_w{batch_id}.csv',
    names=['customer_id', 'product_id']
  )
  pred_pairs = set(map(
    tuple, 
    preds[['customer_id', 'product_id']].values
  ))

  tp = len(pred_pairs & true_pairs)
  fp = len(pred_pairs - true_pairs)
  fn = len(true_pairs - pred_pairs)
  tn = universe_size - tp - fp - fn

  y_true = [1]*tp + [0]*fp + [1]*fn + [0]*tn
  y_pred = [1]*tp + [1]*fp + [0]*fn + [0]*tn

  metrics = {
    'accuracy': accuracy_score(y_true, y_pred),
    'precision': precision_score(y_true, y_pred),
    'recall': recall_score(y_true, y_pred),
    'f1': f1_score(y_true, y_pred),
    'tp': tp,
    'fp': fp,
    'fn': fn,
    'tn': tn
  }

  return metrics
````

````python
def show_batch_performance(
  batch: int,
  accuracy: float, 
  precision: float, 
  recall: float, 
  f1: float,
  tp: int,
  fp: int,
  fn: int,
  tn: int
) -> None:
  """
  Defines a standard way to show batch results in stdout.

  Parameters
  ----------
  batch : int
    The batch identifier.
  accuracy : float
    The model accuracy.
  precision : float
    The model precision.
  recall : float
    The model recall.
  f1 : float
    The model F1-Score.
  tp : int
    True positives.
  fp : int
    False positives.
  fn : int
    False negatives.
  tn : int
    True negatives.
  """
  print(f'-- Results for batch {batch} --')
  print(f'Accuracy: {accuracy:.3f}')
  print(f'Precision: {precision:.3f}')
  print(f'Recall: {recall:.3f}')
  print(f'F1-Score: {f1:.3f}')
  print(f'TP: {tp}')
  print(f'FP: {fp}')
  print(f'FN: {fn}')
  print(f'TN: {tn}')
````

````python
acc_scores = []
prec_scores = []
rec_scores = []
f1_scores = []

scores = [
  acc_scores,
  prec_scores,
  rec_scores,
  f1_scores
]
````

````python
batch = 1
perf = get_batch_performance(batch)
for lst, val in zip(scores, list(perf.values())[:-4]):
  lst.append(val)

show_batch_performance(batch, *perf.values())
````

> **Respuesta**: El modelo usado fue LightGBM con hiperparámetros optimizados, y predijo $3.600$ pares cliente-producto para la primera semana de enero de 2025, para $\lvert C \rvert \cdot \lvert P \rvert = 1.523.499$ combinaciones posibles (es decir, predijo que aproximadamente el $\sim 0\text{,}24\,\%$ de las combinaciones terminaban en compra). Al compararlas con las compras efectivas, se registraron $898$ verdaderos positivos, $2.702$ falsos positivos, $2.366$ falsos negativos y $1.517.533$ verdaderos negativos (en total, $3.264$ compras reales). Con estos valores, el modelo es capaz de detectar el $27\text{,}5\,\%$ de las compras reales, sin embargo, aproximadamente $3$ de cada $4$ compras detectadas no es real ($\sim 75\,\%$). El alto valor de la métrica _Accuracy_ ($\sim 99\text{,}7\,\%$) se debe a la dominancia de la clase negativa.
>
> Si consideramos el modelo que predice aleatoriamente, asignando la clase negativa con probabilidad $\frac{\text{Obs. -}}{\text{Obs. -} ~ + ~ \text{Obs. +}}$ y la clase positiva con probabilidad $\frac{\text{Obs. +}}{\text{Obs. -} ~ + ~ \text{Obs. +}}$, el valor esperado de $\text{F1-Score}$ será $0\text{,}002$. De esta forma, el modelo asociado a este _batch_ supera la estrategia aleatoria (_baseline_).
>
> En este punto, la estrategia más importante fue mejorar la etapa de _Feature Engineering_, agregando $3$ variables asociadas al historial del cliente: 
> - `items_lag_1`: qué cantidad el cliente compró de un producto la semana pasada, para cada semana.
> - `items_rolling_mean_4w`: la cantidad promedio que el cliente compró de un producto, tomando $4$ semanas anteriores a la actual (o las que más se puedan hasta llegar a $4$, que aplica sólo cuando no hay suficiente historial pasado).
> - `purchased_lag_1`: determina si el cliente compró ($1$) o no ($0$) un producto determinado, la semana pasada a la actual.
>
> Sin embargo, también se realizó el preprocesamiento de las entregas anteriores (1 y 2), y se desarrolló la versión inicial del _pipeline_ de Airflow que define el flujo de trabajo para generar la predicción. Dentro de este trabajo previo se encontraba: implementación de _pipelines_ con `SimpleImputer` por la mediana y escalador `MinMaxScaler` para variables numéricas, `SimpleImputer` con la categoría más frecuente y codificador `OneHotEncoder` con `handle_unknown='ignore'` para las variables categóricas, y `DropConstantFeatures` con una tolerancia del $85\,\%$ para todo el conjunto de datos, es decir, no tolera que hayan columnas que repiten el mismo valor más que esa cantidad de veces. Otros aspectos se mencionan en los siguientes párrafos.
>
> Desde la perspectiva del preprocesamiento, fue trascendental reducir la cantidad de clientes y productos, eliminando aquellos que nunca compraron (o fueron comprados _resp._), porque sino, los tiempos de cómputo se disparaban considerablemente (alrededor de horas e incluso días), dado que habían muchas entradas después de hacer el cruce entre cliente, producto y semana del año.
>
> También, fue necesario eliminar las entradas con una cantidad de ítems negativa, porque no era consistente con la vida real. Lo que se hizo fue identificar la llave candidata $(x_\text{product\_id}, x_\text{order\_id})$, porque no se pueden duplicar productos en ordenes (en el peor caso, debería aparecer el producto en cuestión con una cantidad no unitaria), y luego, se agruparon todas las entradas por dicha llave. Después de la agrupación, se tomó el máximo de las modas por grupo, para reducir todos los valores negativos a un valor. Si el máximo de las modas era negativo, dicho valor se filtraba _a posteriori_.
>
> Para el siguiente _batch_, se había pensado cambiar la estrategia de balance de clases, dado que la proporción $1:42$ era muy agresiva, especialmente pensando en que son muchas entradas. Se propusieron opciones como _upsampling_ (SMOTE, ADASYN, etc.), _downsampling_ (NearMiss, Tomek Links, etc.) o modificación del preprocesamiento. Esto fue decidido en la siguiente sección.

> **Respuesta**: El modelo utilizado siguió siendo LightGBM con hiperparámetros optimizados. En este punto, se ve el primer gran salto en la métrica de evaluación $\text{F1-Score}$, pasando de $\approx 0\text{,}26$ a casi el doble, $\approx 0\text{,}52$, escalando a la primera posición en Codabench, y siendo el primer equipo en superar el umbral de $0\text{,}5$.
>
> Siguiéndose de lo anterior, este modelo es mucho mejor que el modelo _baseline_ basado en la estadística frecuentista de los datos (en específico, $260$ veces mejor), y se ve que ahora predice $2.390$ verdaderos positivos, $1.734$ falsos positivos, $2.713$ falsos negativos, y $1.516.662$ verdaderos negativos (en total, $5.103$ compras reales).
>
> La mejora se dio tanto en la métrica _Recall_ como en _Precision_, y _Accuracy_ se mantuvo constante por los motivos antes mencionados. Esta mejora se interpreta como que el modelo es capaz de detectar $\approx 47\,\%$ de las compras reales ($+ 19\text{,}5\,\%$), y además, $58$ de cada $100$ veces que dice que alguien compró, es cierto ($\approx 58\,\%$). De estos resultados, se puede asegurar que el modelo ahora está mucho más seguro, incluso si tenía muy pocos ejemplos positivos para aprender. 
>
> Como consecuencia, las probabilidades de asignación de clase empezaron a tender a los extremos del intervalo $[0, 1]$ (i. e., el modelo le asigna una probabilidad $p \to 0$ a la mayoría de los clientes que no compran cierto producto, y $p \to 1$ si lo hacen), sin embargo, aún no son del todo perfectas. Esto toma especial relevancia, porque en este _batch_ se implementó la **calibración del umbral** que maximizaba la métrica $\text{F1}$. Para ello, hay que considerar que el umbral por defecto de LightGBM es $t = 0\text{,}5$. Si $p(X) \ge t$, el ejemplo es clasificado a la clase positiva. El umbral óptimo es aquel que maximiza la métrica $\text{F1}$, es decir,
> $$
> t^\star = \arg \max_{t} \text{F1}\left(\vec{y}, \vec{\hat{y}} = p_t(\vec{x})\right); \quad p_t(x) = \begin{cases} 1 & \text{si } p(x) \ge t \\ 0 & \text{si no} \end{cases}
> $$
> La implementación de esta estrategia tuvo un impacto de mejora marginal (alrededor de $2$ o $3$ centésimas en $\text{F1}$) con respecto a la mejora total.
>
> Lo que más permitió que el modelo mejorara en este punto fue el **tratamiento del desbalance de clases**. En particular, se verificó que para el _batch_ anterior, el cruce entre clientes y productos para generar el conjunto de entrenamiento estaba siendo completo. Esto es negativo para el rendimiento, porque el modelo aprende patrones irrelevantes. Para ejemplificar esto, se recomienda pensar en una persona natural que tiene sus compras. Es cierto que a veces existen patrones irregulares, por ejemplo, cuando se compra un artículo después de mucho tiempo sin comprarlo, o bien, compras por regalos de navidad, pero son _outliers_. Si dicha persona compra productos, muy rara vez comprará algo fuera de esa lista. Bien, esa es la lógica que se aplicó acá: no hay por qué hacer el cruce completo, si sólo es relevante el cruce de clientes con productos que realmente compran. Esta lógica redujo el desbalance de clases a una proporción $\approx 1:6$, lo que permitió que el modelo aprendiera patrones mucho más relevantes, y fue la médula de este avance.
>
> Una preocupación natural que surge de lo anterior es qué pasa si un cliente $c \in C$ compra un producto $p \in P$, cuando esa tupla $(c, p)$ nunca había sido vista en el conjunto de datos. La solución es sencilla, y sigue la estrategia que se implementó en el entrenamiento (concatenación de _datasets_): reentrenar.
>
> Otra de las mejoras que se implementó en este _batch_ fue la estrategia de _Rolling Window Cross-Validation_, que consiste en una ventana de tiempo que se desplaza para definir los conjuntos de entrenamiento y validación en la optimización de hiperparámetros. Para esto, fue necesario ordenar en el preprocesamiento las fechas, porque esta forma de _splitting_ asume que vienen ordenadas de menor a mayor. Esto permitió que las métricas de validación fueran más cercanas a la realidad, dado que el modelo tenía la oportunidad de aprender de varias ventanas temporales y validar con un futuro que podía ser o no incierto (p. ej., considerando periodos donde las transacciones son atípicas, como Navidad). Además, el entrenamiento del modelo final con los hiperparámetros optimizados, luego de finalizado el estudio de Optuna, se hizo con todo el conjunto de datos, sin perder lo que se tenía en validación, para no perder ninguna información que pudiese ser valiosa.

> **Respuesta**: En este _batch_, el rendimiento superó por $2$ centésimas en la métrica de competencia al _batch_ anterior ($\Delta \text{F1} = +0\text{,}02$). Hasta este punto, se seguía estando en el primer lugar en el podio de Codabench, sin embargo, luego de una investigación, se determinó que se estaba llegando a un punto de **rendimientos decrecientes** (_diminishing returns_), es decir, el modelo empleado LightGBM se había exprimido lo más posible.
>
> En particular, las predicciones arrojaron $3.666$ verdaderos positivos, $4.956$ falsos positivos, $1.256$ falsos negativos, y $1.513.621$ verdaderos negativos (en total, $4.922$ compras reales). El modelo fue capaz de detectar el $74\text{,}5\,\%$ de las compras reales ($\approx +27\text{,}5\,\%$), pero ahora, de $100$ veces que dice que alguien compró, "$42\text{,}5$" veces le acierta (para que tenga sentido, se aproxima a $43$, y eso calza con $\approx 43\,\%$, lo que significa una disminución en $\approx 4\text{,}5\,\%$ con respecto al _batch_ anterior). Esta disminución se explica por la baja exploración que tuvo el `TPESampler` de Optuna por limitaciones de tiempo, ya que se usaron $10$ _trials_, que usualmente se asocian a exploraciones aleatorias en vez de explotación de fortalezas.
>
> Para esta etapa, la mejora principal se sustentó en la adición de las variables de negocio LRFM (_Length_, _Recency_, _Frequency_, _Monetary_) al hacer _Feature Engineering_. Principalmente, esta modificación en el preprocesamiento buscaba resolver la ceguera del modelo ante dos clientes que podían no haber comprado la semana pasada, ni hace $4$ semanas (comportamiento capturado por `items_lag_1`, `items_rolling_mean_4w` y `purchased_lag_1` explicadas en el _batch 1_), pero uno era mucho menos errático que el otro (p. ej., compra cada cierto periodo). Se habla de ceguera, porque hasta el _batch_ anterior, para este modelo, ambos clientes tenían características muy similares, cuando en la práctica no era así.

### 📌 Conclusiones y Aprendizajes [1.0 puntos]

> **Respuesta**: El problema abordado, también conocido como _time-series retail_, tiene varias complicaciones que no se comparten con otra clase de proyectos de Machine Learning. Esto permitió concretar muchos aprendizajes, que se complementan con que los datos eran reales y anonimizados, lo que introduce complicaciones que son usuales en ambientes productivos.
>
> En primera instancia, el **desbalance de clases** es algo natural, y <u>se intensifica a gran escala</u>. Esto se debe a que el universo de clientes ($\lvert C \rvert$) y productos ($\lvert P \rvert$) genera un universo de $\lvert C \rvert \cdot \lvert P \rvert$ posibles transacciones en una semana específica (bajo la hipótesis que el producto $\lvert C \rvert \cdot \lvert P \rvert$ es constante, que se cumplía en este proyecto porque tanto $\lvert C \rvert$ como $\lvert P \rvert$ eran invariantes). Lo primero a considerar es que si $\min \{\lvert C \rvert, \lvert P \rvert\} \approx 10.000 = 10^4$, entonces $\lvert C \rvert \cdot \lvert P \rvert \cdot \underbrace{52}_\text{semanas} > \left(\min \{\lvert C \rvert, \lvert P \rvert \}\right)^2 \cdot 52 = O(10^9)$, o sea, existirán como mínimo alrededor de $10^9$ observaciones en un cruce sin filtros (_merge_), lo que genera problemas en cuanto a la complejidad espacial de la implementación.
>
> Esta complejidad espacial no sólo hace que el preprocesamiento y el entrenamiento sean ineficientes en tiempo (en especial, por la optimización de hiperparámetros), sino que también implementa mucho ruido en el espacio de los datos. En palabras simples, si un cliente $c \in C$ nunca compró un producto $p \in P$ en el tiempo de medición de transacciones, y además $w^\star$ es la última semana del conjunto de datos, entonces la probabilidad de que la tupla $(c, p) \in C \times P$ aparezca la semana $w^\star + 1$ sigue siendo muy baja. Ante esto, la **solución** es reducir el cruce a las compras efectivas (cruzar los clientes con los productos que efectivamente compran), lo que permite tener un universo de transacciones con un cardinal mucho menor al original, y por ende, manejable, y si contra todo pronóstico aparece una nueva tupla $(c, p)$, reentrenar permitirá adquirir esta nueva información.
>
> Luego, hay muchos problemas asociados al **entrenamiento**. Entendiendo que en problemas con series de tiempo siempre hay que entrenar con el pasado y _testear_ con el "presente" (o futuro, si se está en producción), no es una buena idea para _retail_ que las ventanas de entrenamiento y validación sean inmóviles. Esto se debe a que hay semanas atípicas en el año (p. ej., Navidad), donde las transacciones cambian rotundamente. En particular, con esto se encontró la estrategia "_Rolling Window Cross-Validation_" que busca mover las ventanas de entrenamiento y validación a medida que el modelo está capturando la información disponible. Con esto, puede lograr lo siguiente: 
> 1. Tener más contexto de semanas que pueden ser atípicas, y
> 2. comparar en validación dos modelos (entendiéndose que son distintos incluso si sus hiperparámetros cambian) para poder saber cuál es mejor.
>
> En tercer lugar, la **optimización de hiperparámetros** fue clave. En especial, hay que tener cuidado con la calibración de los rangos de la grilla de optimización. Uno de los principales problemas que se experimentaron fue generar una grilla que nunca iba a alcanzar hiperparámetros que realmente fueran beneficiosos para el problema. Por ejemplo, aludiendo al desbalance de clases, si `scale_pos_weight` se exploraba en el intervalo $[a, b]$, y el óptimo teórico $s_\text{OPT} = \frac{\lvert - \rvert}{\lvert + \rvert} \not\in [a,b]$, en especial si $s_\text{OPT} \ll a$ o $s_\text{OPT} \gg b$, entonces el rendimiento incluso podía empeorar notablemente. Como aprendizaje, si el modelo ya era suficientemente competitivo (bajo nuestro contexto, un $\text{F1-Score} > 0.5$), el aumentar el número de _trials_ para mejorar la exploración a lo más subía en $2$ o $3$ centésimas el rendimiento en Codabench.
>
> Bajo este mismo contexto, la <u>elección de la métrica a optimizar</u> también cobra mucha relevancia. En particular, esto explica que en los resultados obtenidos para el mejor _batch_ la métrica _Precision_ haya subido y _Recall_ haya bajado (_trade-off_). En primera instancia, se estaba optimizando $\text{F1-Score}$ con la opción $\texttt{macro}$ ($\text{F1}_\texttt{macro} = \frac{\text{F1}_- + \text{F1}_+}{2}$), sin embargo, esto hacía que el modelo operara "cómodamente", ya que predecir todo como negativo permite maximizar $\text{F1}_-$, el puntaje asociado a la clase negativa (mayoritaria), y no se obtenían puntajes óptimos que superaran el umbral $0\text{,}7$, donde $\text{F1}_+$, el objetivo de predicción, fuese mayor que $0\text{,}4$. La solución en este caso fue beneficiar a la clase minoritaria ($+$), cambiando el objetivo a $\texttt{binary}$, donde sólo se considera $\text{F1}_+$.
>
> Por último, lo más importante fue el **preprocesamiento**. Esto incluye principalmente la etapa de _Feature Engineering_. Lo complejo de esta etapa fue evitar el _Data Leakage_, pues cualquier mínima transformación que filtrara información de una semana futura al conjunto de entrenamiento invalidaba completamente los resultados. Para tratar esto, los mayores aliados fueron los _shifts_ (o _lags_), y la exclusión de variables directamente relacionadas con el _target_ $y_{cpw} \in \{0, 1\}$ (el cliente $c \in C$ compra el producto $p \in P$ en la semana $w \in \{1, \dots, 52\}$ o no). 
>
> Como aprendizaje de esta etapa, las <u>métricas de negocio</u> LRFM (_Length_, _Recency_, _Frequency_, _Monetary_) permiten distinguir al modelo entre un cliente "activo" y uno que nunca más compró. Esto toma especial relevancia cuando se aplican sólo _lags_: si en el periodo comprendido entre la última semana disponible en los datos y la más antigua capturada por los _lags_, los clientes $c_1$ y $c_2$, potenciales compradores del producto $p$, no compraron dicho producto, pero $c_1$ lo compró justo la semana anterior que no fue capturada y $c_2$ no lo compra hace mucho tiempo, el modelo no podrá distinguir entre $c_1$ y $c_2$ ("ceguera de predicción"), siendo que la probabilidad de que $c_1$ compre $p$ debiese ser mayor.
