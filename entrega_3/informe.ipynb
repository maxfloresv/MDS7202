### 📌 Análisis individual [3.0 puntos]

````python
import pandas as pd
from sklearn.metrics import (
  accuracy_score, 
  precision_score, 
  recall_score, 
  f1_score
)

DATA_DIR = './data'
PREDS_DIR = './predictions'

clients = pd.read_parquet(f'{DATA_DIR}/clientes.parquet')
products = pd.read_parquet(f'{DATA_DIR}/productos.parquet')
# Even if some clients and products were eliminated from the train dataset, they
# are good (or bad) predictions too, because they were part of the original data.
universe_size = len(clients) * len(products)

def get_batch_performance(batch_id: int) -> dict:
  """
  Gets the performance for a specific batch.

  Parameters
  ----------
  batch_id : int
    The batch to evaluate.

  Returns
  -------
  dict
    The metrics for the evaluated batch.
  """
  data = pd.read_parquet(f"{DATA_DIR}/batch_t{batch_id}.parquet")
  data = data.groupby(['customer_id', 'product_id']).size().reset_index()
  true_pairs = set(map(
    tuple, 
    data[['customer_id', 'product_id']].values
  ))

  preds = pd.read_csv(
    f'{PREDS_DIR}/preds_w{batch_id}.csv',
    names=['customer_id', 'product_id']
  )
  pred_pairs = set(map(
    tuple, 
    preds[['customer_id', 'product_id']].values
  ))

  tp = len(pred_pairs & true_pairs)
  fp = len(pred_pairs - true_pairs)
  fn = len(true_pairs - pred_pairs)
  tn = universe_size - tp - fp - fn

  y_true = [1]*tp + [0]*fp + [1]*fn + [0]*tn
  y_pred = [1]*tp + [1]*fp + [0]*fn + [0]*tn

  metrics = {
    'accuracy': accuracy_score(y_true, y_pred),
    'precision': precision_score(y_true, y_pred),
    'recall': recall_score(y_true, y_pred),
    'f1': f1_score(y_true, y_pred)
  }

  return metrics
````

````python
def show_batch_performance(
  batch: int,
  accuracy: float, 
  precision: float, 
  recall: float, 
  f1: float
) -> None:
  """
  Defines a standard way to show batch results in stdout.

  Parameters
  ----------
  batch : int
    The batch identifier.
  accuracy : float
    The model accuracy.
  precision : float
    The model precision.
  recall : float
    The model recall.
  f1 : float
    The model F1-Score.
  """
  print(f'-- Results for batch {batch} --')
  print(f'Accuracy: {accuracy:.3f}')
  print(f'Precision: {precision:.3f}')
  print(f'Recall: {recall:.3f}')
  print(f'F1-Score: {f1:.3f}')
````

### 📌 Conclusiones y Aprendizajes [1.0 puntos]

> **Respuesta**: El problema abordado, también conocido como _time-series retail_, tiene varias complicaciones que no se comparten con otra clase de proyectos de Machine Learning. Esto permitió concretar muchos aprendizajes, que se complementan con que los datos eran reales y anonimizados, lo que introduce complicaciones que son usuales en ambientes productivos.
>
> En primera instancia, el **desbalance de clases** es algo natural, y <u>se intensifica a gran escala</u>. Esto se debe al universo de clientes ($\lvert C \rvert$) y productos ($\lvert P \rvert$) genera un universo de $\lvert C \rvert \cdot \lvert P \rvert$ posibles transacciones en una semana específica (bajo la hipótesis que el producto $\lvert C \rvert \cdot \lvert P \rvert$ es constante, que se cumplía en este proyecto porque tanto $\lvert C \rvert$ como $\lvert P \rvert$ eran invariantes). Lo primero a considerar es que si $\min \{\lvert C \rvert, \lvert P \rvert\} \approx 10.000 = 10^4$, entonces $\lvert C \rvert \cdot \lvert P \rvert > \left(\min \{\lvert C \rvert, \lvert P \rvert \}\right)^2 = 10^8$, o sea, existirán más de $10^8$ observaciones en un cruce sin filtros (_merge_), lo que genera problemas en cuanto a la complejidad espacial de la implementación.
>
> Esta complejidad espacial no sólo hace que el preprocesamiento y el entrenamiento sean ineficientes en tiempo (en especial, por la optimización de hiperparámetros), sino que también implementa mucho ruido en el espacio de los datos. En palabras simples, si un cliente $c \in C$ nunca compró un producto $p \in P$ en el tiempo de medición de transacciones, y además $w^\star$ es la última semana del conjunto de datos, entonces la probabilidad de que la tupla $(c, p) \in C \times P$ aparezca la semana $w^\star + 1$ sigue siendo muy baja. Ante esto, la **solución** es reducir el cruce a las compras efectivas (cruzar los clientes con los productos que efectivamente compran), lo que permite tener un universo de transacciones con un cardinal mucho menor al original, y por ende, manejable, y si contra todo pronóstico aparece una nueva tupla $(c, p)$, reentrenar permitirá adquirir esta nueva información.
>
> Luego, hay muchos problemas asociados al **entrenamiento**. Entendiendo que en problemas con series de tiempo siempre hay que entrenar con el pasado y _testear_ con el "presente" (o futuro, si se está en producción), no es una buena idea para _retail_ que las ventanas de entrenamiento y validación sean inmóviles. Esto se debe a que hay semanas atípicas en el año (p. ej., Navidad), donde las transacciones cambian rotundamente. En particular, con esto se encontró la estrategia "_Rolling Window Cross-Validation_" que busca mover las ventanas de entrenamiento y validación a medida que el modelo está capturando la información disponible. Con esto, puede lograr lo siguiente: 
> 1. Tener más contexto de semanas que pueden ser atípicas, y
> 2. comparar en validación dos modelos (entendiéndose que son distintos incluso si sus hiperparámetros cambian) para poder saber cuál es mejor.
>
> En tercer lugar, la **optimización de hiperparámetros** fue clave. En especial, hay que tener cuidado con la calibración de los rangos de la grilla de optimización. Uno de los principales problemas que se experimentaron fue generar una grilla que nunca iba a alcanzar hiperparámetros que realmente fueran beneficiosos para el problema. Por ejemplo, aludiendo al desbalance de clases, si `scale_pos_weight` se exploraba en el intervalo $[a, b]$, y el óptimo teórico $s_\text{OPT} = \frac{\lvert - \rvert}{\lvert + \rvert} \not\in [a,b]$, en especial si $s_\text{OPT} \ll a$ o $s_\text{OPT} \gg b$, entonces el rendimiento incluso podía empeorar notablemente. Como aprendizaje, si el modelo ya era suficientemente competitivo (bajo nuestro contexto, un $\text{F1-Score} > 0.5$), el aumentar el número de _trials_ para mejorar la exploración a lo más subía en $2$ o $3$ centésimas el rendimiento en Codabench.
>
> Bajo este mismo contexto, la <u>elección de la métrica a optimizar</u> también cobra mucha relevancia. En particular, esto explica que en los resultados obtenidos para el mejor _batch_ la métrica _Precision_ haya subido y _Recall_ haya bajado (_trade-off_). En primera instancia, se estaba optimizando $\text{F1-Score}$ con la opción $\texttt{macro}$ ($\text{F1}_\texttt{macro} = \frac{\text{F1}_- + \text{F1}_+}{2}$), sin embargo, esto hacía que el modelo operara "cómodamente", ya que predecir todo como negativo permite maximizar $\text{F1}_-$, el puntaje asociado a la clase negativa (mayoritaria), y no se obtenían puntajes óptimos que superaran el umbral $0\text{,}7$, donde $\text{F1}_+$, el objetivo de predicción, fuese mayor que $0\text{,}4$. La solución en este caso fue beneficiar a la clase minoritaria ($+$), cambiando el objetivo a $\texttt{binary}$, donde sólo se considera $\text{F1}_+$.
>
> Por último, lo más importante fue el **preprocesamiento**. Esto incluye principalmente la etapa de _Feature Engineering_. Lo complejo de esta etapa fue evitar el _Data Leakage_, pues cualquier mínima transformación que filtrara información de una semana futura al conjunto de entrenamiento invalidaba completamente los resultados. Para tratar esto, los mayores aliados fueron los _shifts_ (o _lags_), y la exclusión de variables directamente relacionadas con el _target_ $y_{cpw} \in \{0, 1\}$ (el cliente $c \in C$ compra el producto $p \in P$ en la semana $w \in \{1, \dots, 52\}$ o no). 
>
> Como aprendizaje de esta etapa, las <u>métricas de negocio</u> LRFM (_Length_, _Recency_, _Frequency_, _Monetary_) permiten distinguir al modelo entre un cliente "activo" y uno que nunca más compró. Esto toma especial relevancia cuando se aplican sólo _lags_: si en el periodo comprendido entre la última semana disponible en los datos y la más antigua capturada por los _lags_, los clientes $c_1$ y $c_2$, potenciales compradores del producto $p$, no compraron dicho producto, pero $c_1$ lo compró justo la semana anterior que no fue capturada y $c_2$ no lo compra hace mucho tiempo, el modelo no podrá distinguir entre $c_1$ y $c_2$ ("ceguera de predicción"), siendo que la probabilidad de que $c_1$ compre $p$ debiese ser mayor.
