```python
"""
En la siguiente celda se crean los datos ficticios a usar en la sección 1 del lab.
❗No realice cambios a esta celda a excepción de n_samples❗
"""

n_samples = (1000, 5000, 10000)

def create_data(n_samples):
    # Lunas
    moons = datasets.make_moons(n_samples=n_samples, noise=0.05, random_state=30)
    # Blobs
    blobs = datasets.make_blobs(n_samples=n_samples, random_state=172)
    # Datos desiguales
    transformation = [[0.6, -0.6], [-0.4, 0.8]]
    mutated = (np.dot(blobs[0], transformation), blobs[1])

    # Generamos Dataset
    dataset = {
        'moons':{
            'x': moons[0], 'classes': moons[1], 'n_cluster': 2
        },
        'blobs':{
            'x': blobs[0], 'classes': blobs[1], 'n_cluster': 3
        },
        'mutated':{
            'x': mutated[0], 'classes': mutated[1], 'n_cluster': 3
        }
    }
    return dataset

data_sets = {}
for n in n_samples:
    data_sets[n] = create_data(n)
```

```python
import time
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
from sklearn.mixture import GaussianMixture
from umap import UMAP
from sklearn.metrics import silhouette_score

RANDOM_STATE = 42

projector = UMAP(
    random_state=RANDOM_STATE,
    n_neighbors=20,
    # We search for cluster cohesion 
    min_dist=0.15,
    n_components=2
)

def plot_scatter(x, y, n_clusters, color):
    fig = make_subplots(rows=3, cols=4, subplot_titles=(
        "KMeans", 
        "DBSCAN", 
        "WARD", 
        "GMM"
    ), horizontal_spacing=0.075)

    for i in range(len(x)):
        x_proj = projector.fit_transform(x[i])
        exec_times = {}
        silhouette_scores = {}

        start = time.time()
        kmeans = KMeans(
            n_clusters=n_clusters[i], 
            n_init=10,
            random_state=RANDOM_STATE
        ).fit(x[i])
        exec_times['KMeans'] = time.time() - start
        silhouette_scores['KMeans'] = silhouette_score(x[i], kmeans.labels_)

        fig.add_trace(
            go.Scatter(
                x=x_proj[:, 0], 
                y=x_proj[:, 1], 
                mode='markers', 
                marker=dict(color=kmeans.labels_, colorscale="Viridis"),
            ),
            row=i + 1, col=1
        )

        start = time.time()
        dbscan = DBSCAN(
            eps=0.2, 
            min_samples=5
        ).fit(x[i])
        exec_times['DBSCAN'] = time.time() - start
        silhouette_scores['DBSCAN'] = silhouette_score(x[i], dbscan.labels_)

        dbscan_labels = dbscan.labels_
        mask_clusters = dbscan_labels != -1

        fig.add_trace(
            go.Scatter(
                x=x_proj[mask_clusters, 0],
                y=x_proj[mask_clusters, 1],
                mode="markers",
                marker=dict(
                    color=dbscan_labels[mask_clusters],
                    colorscale="Viridis",
                ),
            ),
            row=i + 1, col=2
        )

        # Noise points from DBSCAN are painted in light gray
        fig.add_trace(
            go.Scatter(
                x=x_proj[~mask_clusters, 0],
                y=x_proj[~mask_clusters, 1],
                mode="markers",
                marker=dict(color="lightgray"),
            ),
            row=i + 1, col=2
        )

        start = time.time()
        ward = AgglomerativeClustering(
            n_clusters=n_clusters[i],
            linkage='ward'
        ).fit(x[i])
        exec_times['WARD'] = time.time() - start
        silhouette_scores['WARD'] = silhouette_score(x[i], ward.labels_)

        fig.add_trace(
            go.Scatter(
                x=x_proj[:, 0], 
                y=x_proj[:, 1], 
                mode='markers', 
                marker=dict(color=ward.labels_, colorscale="Viridis"),
            ),
            row=i + 1, col=3
        )

        start = time.time()
        gmm = GaussianMixture(
            n_components=n_clusters[i], 
            n_init=10,
            random_state=RANDOM_STATE
        ).fit(x[i])
        gmm_labels = gmm.predict(x[i])
        exec_times['GMM'] = time.time() - start
        silhouette_scores['GMM'] = silhouette_score(x[i], gmm_labels)

        fig.add_trace(
            go.Scatter(
                x=x_proj[:, 0], 
                y=x_proj[:, 1], 
                mode='markers', 
                marker=dict(color=gmm_labels, colorscale="Viridis"),
            ),
            row=i + 1, col=4
        )

        for j, method in enumerate(exec_times.keys()):
            fig.add_annotation(
                text=f"{exec_times[method]:.2f} [s] | s: {silhouette_scores[method]:.2f}",
                xref="x domain", 
                yref="y domain", 
                x=0.5, y=1.2, 
                showarrow=False,
                font=dict(size=10),
                row=i + 1, col=j + 1
            )

    for i, title in enumerate(fig.layout.annotations):
        if i < 4:
            title.update(y=title.y + 0.05)

    fig.update_layout(
        width=800,
        height=600,
        showlegend=False, 
        title_text=f"Comparación de técnicas de clustering para {len(x[0])} muestras"
    )
    fig.show()

for n_key in data_sets.keys():
    print(f"Generando gráfico para n={n_key}")
    x = []
    y = []
    n_clusters = []
    for data_key in data_sets[n_key].keys():
        data_value = data_sets[n_key][data_key]
        x.append(data_value['x'])
        y.append(data_value['classes'])
        n_clusters.append(data_value['n_cluster'])
    plot_scatter(x, y, n_clusters, 'blue')
```
