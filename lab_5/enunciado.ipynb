4. Escriba la fórmula de las siguientes métricas y explique con **sus palabras** cómo se interpretan. [1 punto cada uno]

  * Accuracy
  * Precision
  * Recall
  * F1 score

> **Respuesta**: Antes de definir las fórmulas, se necesitan definir los valores $\text{TP}$, $\text{TN}$ para los aciertos: verdaderos positivos y negativos respectivamente, y $\text{FP}$, $\text{FN}$ para los errores: falsos positivos y negativos respectivamente.
> $$
> \text{Accuracy} = \frac{\text{TP} + \text{TN}}{\text{TP} + \text{TN} + \text{FP} + \text{FN}};\, \text{Precision} = \frac{\text{TP}}{\text{TP} + \text{FP}};\, \text{Recall} = \frac{\text{TP}}{\text{TP} + \text{FN}};\, \\[1em] \text{F1-Score} = \frac{2\cdot \text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}
> $$
> Por un lado, $\text{Accuracy}$ mide la proporción de aciertos en las predicciones con respecto al total de clasificaciones, sin diferenciar entre positivas y negativas. Por otro lado, $\text{Precision}$ se enfoca específicamente en cuántas predicciones, de todas las que fueron clasificadas como positivas, se hicieron de manera correcta. Similarmente, $\text{Recall}$ realiza esto, pero con la proporción de ejemplos positivos capturados por el modelo con respecto al total de ejemplos positivos reales. Finalmente, $\text{F1-Score}$ es la media armónica entre $\text{Precision}$ y $\text{Recall}$. Esta media es parecida al promedio, salvo que penaliza mucho más cuando una de las dos métricas es menor, porque usa pesos inversamente proporcionales.

5. Explique qué métrica recomendaría para los siguientes contextos de clasificación. [1 punto cada uno]

  * Mantenimiento predictivo de fallas de maquinaria pesada en la industria minera.

Al revés ($\text{FP}$) puede tener efectos psicológicos, pero _a posteriori_ se puede reevaluar en el seguimiento médico, concluyendo que realmente el paciente no estaba enfermo.
  * Aprobación de créditos de alto riesgo.

  * Detección de crímenes.

3. Genere un `ColumnTransformer` en donde especifique las transformaciones que hay que realizar para cada columna (por ejemplo StandarScaler, MinMaxScaler, OneHotEncoder, etc...) para que puedan ser utilizadas correctamente por el modelo predictivo y guárdelo una variable llamada `col_transformer`.

````python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import (
  StandardScaler, 
  MinMaxScaler, 
  OneHotEncoder,
  OrdinalEncoder
)

std_columns = [
  'Height',
  'Weight',
  'Ball_Control',
  'Dribbling',
  'Aggression',
  'Reactions',
  'Vision',
  'Composure',
  'Short_Pass',
  'Long_Pass',
  'Acceleration',
  'Speed',
  'Stamina',
  'Strength',
  'Balance',
  'Agility',
  'Jumping',
  'Penalties'
]

minmax_columns = [
  'Age',
  'Marking',
  'Sliding_Tackle',
  'Standing_Tackle',
  'Interceptions',
  'Crossing',
  'Heading',
  'Shot_Power',
  'Finishing',
  'Long_Shots',
  'Curve',
  'Freekick_Accuracy',
  'Volleys'
]

onehot_columns = [
  'National_Position',
  'Club_Position',
  'Preffered_Foot',
  'Work_Rate'
]

ordenc_columns = [
  'Weak_foot',
  'Skill_Moves'
]

def create_column_transformer(
  std_columns=std_columns,
  minmax_columns=minmax_columns,
  onehot_columns=onehot_columns,
  ordenc_columns=ordenc_columns
) -> ColumnTransformer:
  """
  Create a column transformer for data preprocessing.

  Parameters
  ----------
  std_columns : list[str]
    List of columns to be standardized using StandardScaler.
  minmax_columns : list[str]
    List of columns to be scaled using MinMaxScaler.
  onehot_columns : list[str]
    List of columns to be one-hot encoded using OneHotEncoder.
  ordenc_columns : list[str]
    List of columns to be ordinally encoded using OrdinalEncoder.

  Returns
  -------
  ColumnTransformer
    A ColumnTransformer that applies the specified transformations to the respective columns.
  """
  ct = ColumnTransformer(
    transformers=[
      ('std_scaler', StandardScaler(), std_columns),
      ('minmax_scaler', MinMaxScaler(), minmax_columns),
      ('onehot_encoder', OneHotEncoder(drop='first', handle_unknown='ignore'), onehot_columns),
      ('ordinal_encoder', OrdinalEncoder(), ordenc_columns)
    ]
  )
  return ct

col_transformer = create_column_transformer()
````

> **Respuesta**: Las transformaciones elegidas se separarán por el procesador utilizado para su justificación.
> - `StandardScaler`: Para variables numéricas que seguían una distribución similar a una normal.
> - `MinMaxScaler`: Para el resto de las variables numéricas que no seguían la distribución mencionada.
> - `OneHotEncoder`: Para variables categóricas que no seguían un orden. Se usó la opción `drop='first'` para evitar multicolinealidad perfecta, pues la suma de todas las variables de diseño dan exactamente $1$ en el caso de no eliminar alguna de estas variables.
> - `OrdinalEncoder`: Para el resto de variables categóricas, que sí seguían un orden (p. ej., una escala del $1$ al $5$ en el caso de `Skill_Moves`).
