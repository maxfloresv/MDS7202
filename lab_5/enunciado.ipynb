4. Escriba la fórmula de las siguientes métricas y explique con **sus palabras** cómo se interpretan. [1 punto cada uno]

  * Accuracy
  * Precision
  * Recall
  * F1 score

> **Respuesta**: Antes de definir las fórmulas, se necesitan definir los valores $\text{TP}$, $\text{TN}$ para los aciertos: verdaderos positivos y negativos respectivamente, y $\text{FP}$, $\text{FN}$ para los errores: falsos positivos y negativos respectivamente.
> $$
> \text{Accuracy} = \frac{\text{TP} + \text{TN}}{\text{TP} + \text{TN} + \text{FP} + \text{FN}};\, \text{Precision} = \frac{\text{TP}}{\text{TP} + \text{FP}};\, \text{Recall} = \frac{\text{TP}}{\text{TP} + \text{FN}};\, \\[1em] \text{F1-Score} = \frac{2\cdot \text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}
> $$
> Por un lado, $\text{Accuracy}$ mide la proporción de aciertos en las predicciones con respecto al total de clasificaciones, sin diferenciar entre positivas y negativas. Por otro lado, $\text{Precision}$ se enfoca específicamente en cuántas predicciones, de todas las que fueron clasificadas como positivas, se hicieron de manera correcta. Similarmente, $\text{Recall}$ realiza esto, pero con la proporción de ejemplos positivos capturados por el modelo con respecto al total de ejemplos positivos reales. Finalmente, $\text{F1-Score}$ es la media armónica entre $\text{Precision}$ y $\text{Recall}$. Esta media es parecida al promedio, salvo que penaliza mucho más cuando una de las dos métricas es menor, porque usa pesos inversamente proporcionales.

5. Explique qué métrica recomendaría para los siguientes contextos de clasificación. [1 punto cada uno]

  * Mantenimiento predictivo de fallas de maquinaria pesada en la industria minera.  
  > **Respuesta**: Maximizar $\text{Recall}$, porque es más grave decir que una máquina no falla cuando realmente sí fallaba ($\text{FN}$) por las consecuencias que eso puede traer en la industria minera, que al revés, donde se dice que la máquina falla, y en el mantenimiento se verá que realmente no tenía ningún problema, entonces se puede descartar.
  * Detección de enfermedades altamente contagiosas.
  > **Respuesta**: Maximizar $\text{Recall}$, porque es mucho más grave un falso negativo. Esto último corresponde a decirle a una persona que no está enferma, cuando realmente sí lo estaba, que podría incluso terminar en el fallecimiento del paciente, o una epidemia. Al revés ($\text{FP}$) puede tener efectos psicológicos, pero _a posteriori_ se puede reevaluar en el seguimiento médico, concluyendo que realmente el paciente no estaba enfermo.
  * Aprobación de créditos de alto riesgo.
  > **Respuesta**: Depende de la perspectiva por dónde se mire, pero en general, desde una visión empresarial, es mucho peor un falso positivo ($\text{FP}$), dado que estaría aprobándole el crédito a una persona que no es apta para poder pagarlo. En un falso negativo ($\text{FN}$), por lo menos el cliente se puede reevaluar con otra entidad que maneje este tipo de créditos. Así, maximizaría $\text{Precision}$.
  * Detección de crímenes.
  > **Respuesta**: Maximizar $\text{F1-Score}$, porque tanto un falso positivo como un falso negativo son graves. En el caso de un falso positivo porque estaríamos incriminando a alguien inocente, y en el caso de un falso negativo porque estaríamos dejando libre a un criminal.

3. Genere un `ColumnTransformer` en donde especifique las transformaciones que hay que realizar para cada columna (por ejemplo StandarScaler, MinMaxScaler, OneHotEncoder, etc...) para que puedan ser utilizadas correctamente por el modelo predictivo y guárdelo una variable llamada `col_transformer`.

````python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import (
  StandardScaler, 
  MinMaxScaler, 
  OneHotEncoder,
  OrdinalEncoder
)

std_columns = [
  'Height',
  'Weight',
  'Ball_Control',
  'Dribbling',
  'Aggression',
  'Reactions',
  'Vision',
  'Composure',
  'Short_Pass',
  'Long_Pass',
  'Acceleration',
  'Speed',
  'Stamina',
  'Strength',
  'Balance',
  'Agility',
  'Jumping',
  'Penalties'
]

minmax_columns = [
  'Age',
  'Marking',
  'Sliding_Tackle',
  'Standing_Tackle',
  'Interceptions',
  'Crossing',
  'Heading',
  'Shot_Power',
  'Finishing',
  'Long_Shots',
  'Curve',
  'Freekick_Accuracy',
  'Volleys'
]

onehot_columns = [
  'National_Position',
  'Club_Position',
  'Preffered_Foot',
  'Work_Rate'
]

ordenc_columns = [
  'Weak_foot',
  'Skill_Moves'
]

def create_column_transformer(
  std_columns=std_columns,
  minmax_columns=minmax_columns,
  onehot_columns=onehot_columns,
  ordenc_columns=ordenc_columns
) -> ColumnTransformer:
  """
  Create a column transformer for data preprocessing.

  Parameters
  ----------
  std_columns : list[str]
    List of columns to be standardized using StandardScaler.
  minmax_columns : list[str]
    List of columns to be scaled using MinMaxScaler.
  onehot_columns : list[str]
    List of columns to be one-hot encoded using OneHotEncoder.
  ordenc_columns : list[str]
    List of columns to be ordinally encoded using OrdinalEncoder.

  Returns
  -------
  ColumnTransformer
    A ColumnTransformer that applies the specified transformations to the respective columns.
  """
  ct = ColumnTransformer(
    transformers=[
      ('std_scaler', StandardScaler(), std_columns),
      ('minmax_scaler', MinMaxScaler(), minmax_columns),
      ('onehot_encoder', OneHotEncoder(drop='first', handle_unknown='ignore'), onehot_columns),
      ('ordinal_encoder', OrdinalEncoder(), ordenc_columns)
    ]
  )
  return ct

col_transformer = create_column_transformer()
````

> **Respuesta**: Las transformaciones elegidas se separarán por el procesador utilizado para su justificación.
> - `StandardScaler`: Para variables numéricas que seguían una distribución similar a una normal.
> - `MinMaxScaler`: Para el resto de las variables numéricas que no seguían la distribución mencionada.
> - `OneHotEncoder`: Para variables categóricas que no seguían un orden. Se usó la opción `drop='first'` para evitar multicolinealidad perfecta, pues la suma de todas las variables de diseño dan exactamente $1$ en el caso de no eliminar alguna de estas variables.
> - `OrdinalEncoder`: Para el resto de variables categóricas, que sí seguían un orden (p. ej., una escala del $1$ al $5$ en el caso de `Skill_Moves`).

````python
from sklearn.model_selection import train_test_split

X = df_players.drop(columns=['label', 'Name', 'Nationality', 'National_Position'])
y = df_players['label']

X_train, X_test, y_train, y_test = train_test_split(
  X, y, test_size=0.3, random_state=42, stratify=y
)
````

2. Defina un pipeline llamado `pipeline_xgboost` y otro llamado `pipeline_lightgbm`. Estos pipelines deben tener el mismo ColumnTransformer definido en la sección de preprocesamiento, pero deben variar los clasificadores de acuerdo al nombre de cada pipeline.

````python
from sklearn.pipeline import Pipeline
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

col_transformer = create_column_transformer(
  onehot_columns=list(filter(lambda x: x != 'National_Position', onehot_columns))
)

pipeline_xgboost = Pipeline(steps=[
  ('preprocessor', col_transformer),
  ('classifier', XGBClassifier())
])

pipeline_lightgbm = Pipeline(steps=[
  ('preprocessor', col_transformer),
  ('classifier', LGBMClassifier(verbose=-1))
])
````

3. Entrene los pipelines.

````python
pipeline_xgboost.fit(X_train, y_train)
pipeline_lightgbm.fit(X_train, y_train)
````

1. Considerando la variable llamada `label` creada en la sección 1.1. Para determinar cuál modelo de árbol sería más adecuado para la tarea en cuestión, utilice PyCaret. Este deberá centrarse exclusivamente en modelos de tipo árbol. Jere ha especificado que busca un modelo que tome decisiones rápidamente y que tenga una baja tasa de falsos positivos, ya que planea invertir en estos jugadores.

> **Respuesta**: Considerando lo justificado en la sección 1.1, se eliminarán las variables `Name`, `Nationality` y `National_Position`. Es muy importante eliminar esta última para que los datos de entrenamiento y prueba no predigan con el patrón trivial revisando por la condición `notna()`.
>
> También, los modelos se entrenarán con un $70\, \%$ de los datos por los motivos ya justificados. El restante $30\, \%$ no se tocará, salvo para evaluar el rendimiento de los modelos, dado que es el conjunto de prueba. Se hizo un _fine tuning_ con `n_iter=5` (5 combinaciones de hiperparámetros), y para evaluar el rendimiento en entrenamiento, se hizo _cross-validation_ con `fold=5` (5 _folds_).
>
> Todos los modelos entrenados fueron guardados en la carpeta `models/` de este laboratorio, con extensión `.pkl`. Posteriormente, serán cargados con `load_model` para ser evaluados en el conjunto de prueba. Las métricas de cada _fold_ están en `models/metrics/`. Se priorizó optimizar Precision para tener una baja tasa de falsos positivos.

````python
from pycaret.classification import *
import os
import gc
import time

os.environ["PYCARET_CUSTOM_LOGGING_LEVEL"] = "CRITICAL"

ct = create_column_transformer(
  onehot_columns=list(
    filter(
      lambda x: x != 'National_Position', onehot_columns
    )
  )
)

df = df_players.drop(
  columns=[
    'Name',
    'Nationality',
    'National_Position', 
    'label'
  ])
df = pd.DataFrame(
  ct.fit_transform(df), 
  columns=ct.get_feature_names_out()
)
df['label'] = df_players['label'].values

X_train, X_test, y_train, y_test = train_test_split(
  df.drop(columns=['label']), 
  df['label'], 
  test_size=0.3, 
  random_state=42, 
  stratify=df['label']
)

train = X_train.copy()
train['label'] = y_train

test = X_test.copy()
test['label'] = y_test

def train_evaluate_model(model_name: str) -> None:
  """
  Train and evaluate a tree classification model using PyCaret, optimizing Precision metric.
  If the tuned model already exists, the function won't retrain it.

  Parameters
  ----------
  model_name : str
    The name of the model to be trained. Supported models for this function are:
    'dt' (Decision Tree), 'rf' (Random Forest), 'et' (Extra Trees), 
    'xgboost' (XGBoost), 'lightgbm' (LightGBM) and 'catboost' (CatBoost).
  """
  if os.path.exists(f'models/tuned_{model_name}.pkl'):
    print(f"Model tuned_{model_name} already exists. Skipping training.")
    return
  
  clf = setup(
    data=train, 
    target='label',
    session_id=42,
    fold=5
  )

  start = time.time()
  if model_name == 'xgboost':
    model = create_model(
      model_name, 
      tree_method='gpu_hist', 
      predictor='gpu_predictor'
    )
  elif model_name == 'catboost':
    model = create_model(
      model_name, 
      task_type='GPU'
    )
  else:
    model = create_model(model_name)
  tuned_model = tune_model(model, n_iter=5, optimize='Precision')
  end = time.time()
  print(f"Training and tuning {model_name} took {end - start:.2f} seconds.")
  save_model(tuned_model, f'models/tuned_{model_name}')

  metrics = pull()
  metrics.to_excel(f"models/metrics/{model_name}.xlsx", index=False)

  del model, tuned_model, metrics, clf
  gc.collect()

model_list = [
  'dt', 
  'rf', 
  'et', 
  'xgboost', 
  'lightgbm', 
  'catboost'
]

for model in model_list:
  train_evaluate_model(model)
````

> - `dt` (Decision Tree): es un árbol de decisión tradicional. En cada nodo, elige una de las variables disponibles para cortar el _dataset_ en $2$ partes, priorizando aquellas variables que maximicen la calidad del corte, con las métricas ya definidas en el método `et`. La predicción $y^\ast$ del árbol en clasificación para una muestra $x^\ast$ es la hoja a la cual llega en la estructura.

> _Nota_: En ambos casos, tanto para `et` como para `rf`, si hay empate en la votación para dos clases $y_1^\ast$ e $y_2^\ast$, hay métodos de desempate, por ejemplo, realizando una ponderación distinta por el voto que emite cada árbol.

1. Se solicita que utilice un método de calibración que asegure que las probabilidades generadas incrementen de manera monótona. Una métrica ampliamente utilizada para evaluar la precisión de la calibración de un modelo es el Brier Score. Calcule el Brier Score para el modelo tanto antes como después de la calibración.

> **Respuesta**: Considerando que el _Brier Score_ se define con $N$ predicciones, y además $f_t$ es la probabilidad de que ocurra el suceso, y $o_t$ el valor observado, la relación matemática es la siguiente:
> $$
> \textit{Brier Score} = \frac{1}{N} \sum_{t=1}^N (f_t - o_t)^2
> $$
> de lo cual se deduce que a menor _Brier Score_, mejor es la predicción realizada por el modelo, que en este caso es XGBoost con reducción de dimensionalidad.
>
> Cabe a destacar que se utiliza `model.predict_proba(X_test)[:, 1]` ya que esto devuelve el arreglo unidimensional de probabilidades predichas por el modelo para la clase positiva, que permite calcular la curva de calibración y el _Brier Score_.

````python
from sklearn.calibration import CalibratedClassifierCV
from sklearn.metrics import brier_score_loss
from sklearn.calibration import calibration_curve

y_pred_reduced = model_xgb_reduced.predict_proba(X_reduced_test)[:, 1]
brier_score = brier_score_loss(y_reduced_test, y_pred_reduced)
print(f'Brier Score sin calibración: {brier_score}')
````

> Dado que el método _Platt Scaling_ asume una forma sigmoidal, la cual no podemos afirmar fácilmente en los datos, se utilizará un método no paramétrico llamado _Isotonic Regression_, el cual permite que se asegure lo pedido: que las probabilidades generadas incrementen monótonamente, ya que busca ajustar una función monótona no decreciente a los datos, mientras que _Platt Scaling_ ajusta a una forma sigmoidal, lo cual no garantiza la monotonía en las probabilidades, ya que es más sensible al ruido y desorden en los datos, cosa que _Isotonic Regression_ maneja mejor.

````python
calibrate_model = CalibratedClassifierCV(
  estimator=model_xgb_reduced, 
  method='isotonic',
  cv=5
)
calibrate_model.fit(X_reduced_train, y_reduced_train)
y_calibrated = calibrate_model.predict_proba(X_reduced_test)[:, 1]
brier_score_calibrated = brier_score_loss(y_reduced_test, y_calibrated)
print(f'Brier Score con calibración: {brier_score_calibrated}')
````
