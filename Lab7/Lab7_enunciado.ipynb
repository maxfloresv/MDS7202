```python
import plotly.express as px
import plotly.graph_objects as go
from scipy import stats

variables = ['MentHlth', 'PhysHlth', 'GenHlth','Age', 'BMI']
for var in variables:
    fig = px.histogram(df, x=var, title=f'Distribución de {var} vs Normal', 
                       nbins=30, histnorm='probability density')
    mean_val = df[var].mean()
    std_val = df[var].std()
    x_vals = np.linspace(df[var].min(), df[var].max(), 100)
    y_vals = stats.norm.pdf(x_vals, mean_val, std_val)
    fig.add_trace(go.Scatter(x=x_vals, y=y_vals, mode='lines', 
                            name='Distribución Normal', line=dict(color='red', width=3)))
    fig.show()
```

> **Respuesta** : Es posible notar con los gráficos, que la distribución de las variables no coinciden del todo con distribuciones normales, por tanto, se debe descartar utilizar _StandardScaler_, para nivelarlas con respecto al resto de las variables, se utilizará _MinMaxScaler_ en ambas variables.

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import MinMaxScaler
from sklearn.pipeline import Pipeline
from xgboost import XGBClassifier
from sklearn.metrics import classification_report
```

```python
train, test= train_test_split(df, test_size=0.33, random_state=42)
train.shape, test.shape
X_train = train.drop(columns=['Diabetes'])
y_train = train['Diabetes']
X_test = test.drop(columns=['Diabetes'])
y_test = test['Diabetes']
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('scaler', MinMaxScaler(), variables),
    ],
    remainder='passthrough',
    verbose_feature_names_out=False
).set_output(transform='pandas')
```

```python
pipeline=Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', XGBClassifier())
])
pipeline.fit(X_train, y_train)
```

```python
y_pred = pipeline.predict(X_test)
print(classification_report(y_test, y_pred))
```

>
> Las métricas utilizadas en _classification report_ buscan metrizar distintas proporciones:
> - Precision: ¿Cuántas personas que se predijo tenían diabetes genuinanamente tenían?
> - Recall: ¿Cuántos casos de los que realmente tenían diabetes se detectaron correctamente?
> - F1-Score: ¿Cómo se equilibran _precision_ y _recall_ entre sí?, es decir, ¿qué tan preciso y acertado es el modelo?
> Las métricas utilizadas permiten entender que tantos casos predijo de la variable objetivo, sin embargo, no nos entrega información sobre las características en las cuales basó su decisión y su forma de decidir, lo cual es importante, ya que _XGBoost_ es un modelo basado en árboles.
