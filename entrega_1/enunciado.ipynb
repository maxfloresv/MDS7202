A continuación, se buscará la **llave candidata**, es decir, la superllave minimal. Para las tablas `customers` y `products`, es intuitivo que estas llaves corresponden a `customer_id` y `product_id` respectivamente. Esto se puede confirmar con las siguientes condiciones _booleanas_.

Para `customers`:

Sin embargo, para `transactions` el tema es más difícil: ya no es intiutivo decir que `order_id` es una superllave, pues pueden haber múltiples transacciones que se generaron en una orden. Esto se confirma en la siguiente celda de código.

```python
transactions['order_id'].nunique() == transactions.shape[0]
```

Acá, es más intuitivo que la llave candidata sea la tupla $($`product_id`$,$ `order_id`$)$, dado que las otras $2$-combinaciones posibles generan repeticiones: `order_id` con `customer_id` se puede repetir en el caso de que un cliente compre varios productos en una misma orden, y `customer_id` con `product_id` se puede repetir cuando un cliente compra el mismo producto en distintas órdenes.

A continuación para confirmar su naturaleza de llave candidata, se corroborará si existen duplicados de la combinación de `product_id` y `order_id`.

```python
key_cols = ["product_id", "order_id"]
```

```python
duplicated = (
    transactions.groupby(key_cols)
        .size()
        .reset_index(name="count")
)
duplicated = duplicated[duplicated['count'] > 1]
duplicated = duplicated.sort_values("count", ascending=False)
duplicated.head(10)
```

Dada la existencia de duplicados para la tupla candidata, se procederá a revisar el caso particular de la fila que tiene más duplicados.

```python
transactions[
    (transactions["product_id"] == 11262) 
    & (transactions["order_id"] == 1210079)
]
```

Al analizar el caso particular, se puede ver que existen filas donde los ítems son negativos, contrastando con el hecho de que el número de ítems debe ser positivo intuitivamente.

A continuación, se calcularán las observaciones que tienen ítems con cantidad negativa y son únicos, es decir, no hay otra observación con la misma _pseudollave candidata_ que tenga otra cantidad de ítems.

```python
neg_items_projection = transactions[
    transactions['items'] < 0
][['product_id', 'order_id']]

dupes_projection = duplicated[['product_id', 'order_id']]

neg_items_rows = neg_items_projection.shape[0]

set_neg = set(neg_items_projection.apply(tuple, axis=1))
set_dupes = set(dupes_projection.apply(tuple, axis=1))

diff_set = set_neg - set_dupes
diff_set
```

```python
for dtuple in diff_set:
    product_id, order_id = dtuple
    unique_neg = transactions[
        (transactions['product_id'] == product_id)
        & (transactions['order_id'] == order_id)
    ]
    print(unique_neg)
    print("-" * 10)
```

Tomando una fila $\vec{x} \in \mathbb{R}^d$ que tiene una combinación de valores $(x_\texttt{product\_id}, x_\texttt{order\_id})$ única, entonces el porcentaje de filas con ítems menores que $0$ donde **no hay otra observación asociada a la misma combinación** se calcula como $\lvert N \setminus D \rvert / \lvert N \rvert$, donde $N$ es el conjunto de observaciones $\vec{x} \in \mathbb{R}^d$ tales que $x_\texttt{items} < 0$, y $D$ es el conjunto de observaciones $\vec{x} \in \mathbb{R}^d$ tales que $\exists \vec{x}' \in \mathbb{R}^d$, $\vec{x} \neq \vec{x}'$, $x_K = x'_K$, con $K$ el subconjunto de variables asociadas a la llave candidata $(x_\texttt{product\_id}, x_\texttt{order\_id})$.

En la siguiente celda de código, se puede verificar el siguiente resultado:
$$
    \frac{\lvert N \setminus D \rvert}{\lvert N \rvert} \approx 0\text{,}08\, \%
$$

```python
print(f'{(len(diff_set) / neg_items_rows) * 100:.2f} %')
```

Por otro lado, la proporción de observaciones con ítems negativos con respecto al total de transacciones se puede definir como $\lvert N \rvert / \lvert T \rvert$, donde $T$ es el conjunto de transacciones. En la siguiente celda de código, se puede verificar que esa división da como resultado $3\text{,}29\,\%$, es decir, un poco más del $3\,\%$ de las observaciones tienen ítems negativos.

```python
print(f'{(neg_items_rows / transactions.shape[0]) * 100:.2f} %')
```

Las observaciones que pertenecen al conjunto $N \setminus D$ se van a considerar como pedidos fallidos, dado que no existe otra observación con una combinación $(x_\texttt{product\_id}, x_\texttt{order\_id})$ asociada a la _pseudollave candidata_ (_pseudo_ porque se vio que hay identificadores repetidos), entonces no hay información sobre cuántos ítems se llevaron realmente, pues debiese ser una cantidad mayor que $0$, como ocurre en las observaciones que viven en $N \cap D$.

```python
mask = (
    transactions[['product_id', 'order_id']]
        .apply(tuple, axis=1)
        .isin(diff_set)
)
print("Filas a eliminar:", transactions[mask].shape[0])
transactions = transactions[~mask]
```

El único conjunto de observaciones con ítems negativos que queda por analizar es $N \cap D$, dado que:
$$
    N = (N \setminus D) \cup (N \cap D); \quad (N \setminus D) \cap (N \cap D) = \emptyset
$$
Para reducir las observaciones con _pseudollave candidata_ duplicada, se usará el criterio del máximo de las modas, es decir:
$$
        C_{p,q} =  \{\vec{x} \in \mathbb{R}^d \mid x_\texttt{product\_id} = p \land x_\texttt{order\_id} = q \} \mapsto \{\vec{x}'\} \text{ tal que } x'_\texttt{items} = \max \text{moda}(C^\texttt{items}_{p, q})
$$
donde $C^\texttt{items}_{p,q}$ es el multiconjunto $\{\pi_\texttt{items}(\vec{x}) \mid \vec{x} \in C_{p,q} \}$, y $\pi_{(\cdot)}$ es la proyección de un vector a una de sus características. Es importante notar que la moda pueden ser varios valores, por eso se toma el más grande de ellos.

Esto se justifica con que en una orden mal ingresada, como es el caso de estas observaciones duplicadas con ítems negativos, el valor que más se repite tiende a ser el correcto.

```python
prev_rows = transactions.shape[0]
mode_max = (
    transactions
        .groupby(['product_id', 'order_id'])['items']
        .transform(lambda x: x.mode().max())
)
mode_max
```

```python
transactions = transactions[
    transactions['items'] == mode_max
]
transactions.drop_duplicates(inplace=True)
print(f"Se eliminaron {prev_rows - transactions.shape[0]} filas")
```

A continuación, se verificará que no queden transacciones con una cantidad de ítems negativa:

```python
transactions[transactions['items'] < 0]
```

A continuación, se eliminarán los _customers_ que no tienen ninguna transacción, y los productos que nunca se han comprado, para limpiar el _dataset_ de entradas irrelevantes.

```python
def get_filtered_customers(min_transactions: int = 1) -> pd.DataFrame:
  """
  Get customers with at least a minimum number of transactions.

  Parameters
  ----------
  min_transactions : int
    Minimum number of transactions a customer must have to be included.

  Returns
  -------
  pd.DataFrame
    Filtered customers DataFrame.
  """
  counts = transactions['customer_id'].value_counts()
  filtered_customers_id = counts[counts >= min_transactions].index
  return customers[customers['customer_id'].isin(filtered_customers_id)]

def get_filtered_products(min_transactions: int = 1) -> pd.DataFrame:
  """
  Get products bought at least a minimum number of times.

  Parameters
  ----------
  min_transactions : int
      Minimum number of transactions a product must have to be included.

  Returns
  -------
  pd.DataFrame
      Filtered products DataFrame.
  """
  counts = transactions['product_id'].value_counts()
  filtered_products_id = counts[counts >= min_transactions].index
  return products[products['product_id'].isin(filtered_products_id)]

print("Antigua shape de clientes:", customers.shape[0])
filtered_customers = get_filtered_customers()
customers = customers[
  customers['customer_id']
    .isin(
      filtered_customers['customer_id']
    )
]
print("Nueva shape de clientes:", customers.shape[0])
print("Antigua shape de productos:", products.shape[0])
filtered_products = get_filtered_products()
products = products[
  products['product_id']
    .isin(
      filtered_products['product_id']
    )
]
print("Nueva shape de productos:", products.shape[0])
```

Para unificar las tablas, se debe considerar el objetivo, que es predecir cada semana la probabilidad de que cada cliente compre cada producto del catálogo. Entonces, la tabla final debiese contener observaciones donde haya un cliente asociado a los $N$ productos, todas las semanas del año enumeradas con la función `isocalendar()`, y si compró o no dicha semana.

```python
transactions['week'] = (
    transactions['purchase_date'].dt.isocalendar().week
)

customer_ids = customers['customer_id'].unique()
product_ids = products['product_id'].unique()
weeks = transactions['week'].unique()

base = pd.MultiIndex.from_product(
    [customer_ids, product_ids, weeks],
    names=['customer_id', 'product_id', 'week']
).to_frame(index=False)

print(f"Observaciones del cruce:", base.shape[0])
base.head()
```

```python
# Esto indica que compró en una semana específica
weekly_trx = (
    transactions.groupby(['customer_id', 'product_id', 'week'])
        .agg({'items': 'sum'})
        .reset_index()
)

df = base.merge(
    weekly_trx[['customer_id', 'product_id', 'week', 'items']],
    on=['customer_id', 'product_id', 'week'],
    how='left'
)

df.head()
```

Considerando esta fusión, se define la siguiente variable objetivo (`label`):
$$
    y_{i,j,t} = \begin{cases}
        1 & \text{si el cliente } i \text{ compró el producto } j \text{ en la semana } t \\
        0 & \text{si no}
    \end{cases}
$$

donde $i$ se extrae de `customer_id`, $j$ de `product_id` y $t$ de `week`.

Esta variable se buscará predecir mediante métodos de _Machine Learning_ supervisado que estimen la probabilidad de la siguiente manera:
$$
    \hat{y}_{i,j,t} = \mathbb{P}(y_{i,j,t} = 1 \mid \vec{x}_{i,j,t})
$$
donde $\vec{x}_{i,j,t}$ es el vector de características asociado a la observación $(i,j,t)$, es decir, la probabilidad que el cliente $i$ compre el producto $j$ en la semana $t$ dadas sus características como cliente.

Para evaluar correctamente el modelo y garantizar su capacidad de generalización, se deben dividir los datos en tres conjuntos: 
- `Entrenamiento` : Para ajustar los parámetros.
- `Validación`: Para optimizar hiperparámetros y seleccionar el mejor modelo.

👀 **Hint**: *Recuerde que los datos tienen una temporalidad que debe considerarse al momento de separarlos, para evitar fugas de información. Es importante justificar la estrategia de partición elegida y visualizar la distribución temporal de los conjuntos generados*

Dada la temporalidad de los datos, es crucial considerar los conjuntos por separado de manera secuencial para no romper su naturaleza temporal, no se debe aleatorizar ya que al aleatorizar, como lo seria un `train_test_split` tradicional de `sklearn`, se perdería la granularidad de la información y estariamos rompiendo la serie de tiempo creada con los procesos anteriores.

Con base en esto y los conocimientos del curso, al ser $52$ semanas de datos, se propone un split secuencial de $70\,\%-15\,\%-15\,\%$ donde se definen los siguientes conjuntos de datos:

`train`: Datos de entrenamiento correspondientes a las compras desde la semana 1 a 36 del año, correspondiente al $70\%$ aproximadamente.

`validation`: Datos de validación correspondientes a las semanas 37 a 44 del año, correspondiente al $15\%$, esto corresponde a la naturaleza de series de tiempo de los datos.

`test`: Datos de prueba correspondientes a la semanas 45 a 52 del año, nuevamente el $15\%$.

Estos conjuntos cumplen ser partición del conjunto original de semanas, sin intersecciones que puedan generar _data leakage_ y mantienen la naturaleza temporal del _dataframe_.

Pese a que no necesariamente posean la misma distribución, la información entregada en entrenamiento al ser $70\%$ debería poder entregar el poder suficiente de generalización al modelo para trabajar correctamente en los otros conjuntos, esta división se respalda en el contenido del curso, donde se menciona esta división para casos donde existe validación y optimización de hiperparámetros.

A continuación se realizará la partición dentro de `processed_df`

```python
train_end = 36 
val_end = 44 
test_end = 52 

train = processed_df[
  processed_df['week'] <= train_end
].copy()
validation = processed_df[
  (processed_df['week'] > train_end) 
  & (processed_df['week'] <= val_end)
].copy()
test  = processed_df[
  (processed_df['week'] > val_end) 
  & (processed_df['week'] <= test_end)
].copy()
```

Para hacer la clasificación, cada árbol vota por la clase que decidió, y se clasifica una muestra nueva $x^\ast$ según la clase más votada $y^\ast$ creando el par asociativo $(x^\ast, y^\ast)$. Sus hiperparámetros más importantes son: `n_estimators` (número de árboles que votan), `max_depth` (profundidad máxima de cada árbol), `min_samples_split` y `min_samples_leaf` (número mínimo de muestras requerida por corte y por hoja), `max_features` (número máximo de variables consideradas por árbol), `class_weight` (peso de cada clase) y `criterion` (métrica de impureza).

Bajo esta definición, `XGBoost` está hecho para que cada árbol reduzca el gradiente de una función de pérdida con respecto a las predicciones anteriores, añadiendo una regularización explícita. Los árboles crecen por niveles, de manera equilibrada, es decir, cada nodo interno del árbol tiene una expansión, aún así no sea informativa. La clasificación consiste en aplicar una función de activación (usualmente, la función _sigmoide_) sobre la suma de todas las contribuciones de las decisiones. Eso otorga un puntaje en $[0, 1]$ que es transformado a una clase. Sus hiperparámetros clave son `n_estimators`, `max_depth`, ya explicados, `learning_rate` (cuánto contribuye cada árbol nuevo), `min_child_weight` (valor mínimo para la suma de contribuciones en una hoja), `gamma` (mínima ganancia requerida para hacer una división), `subsample` (porcentaje de muestras usadas por árbol), `colsample_bytree` (fracción de _features_ usadas por árbol), `scale_pos_weight` (le da más importancia a los errores de la clase positiva) y `reg_alpha`, `reg_lambda` (regularizaciones $L_1$ y $L_2$).

La principal diferencia con `XGBoost` radica en que ahora el crecimiento es por hojas, es decir, se expanden las hojas que aportan mayor ganancia de información. Sus hiperparámetros más importantes son: `num_leaves` (número máximo de hojas por árbol), `max_depth` (ya explicado), `min_data_in_leaf` (número mínimo de datos por hoja), `learning_rate` y `n_estimators` (ya explicados), `feature_fraction` (porcentaje de _features_ por árbol), `bagging_fraction`, `bagging_freq` (cómo se submuestrean las instancias), `lambda_l1`, `lambda_l2` (regularizaciones $L_1$ y $L_2$) y `is_unbalance`, `scale_pos_weight` (ajuste de pesos de clase).

Al desarrollar este proceso, deberán responder las siguientes preguntas clave como mínimo:

- ¿Qué métrica decidieron optimizar y por qué?

- ¿Qué hiperparámetro tuvo un mayor impacto en el rendimiento de su modelo?

> **Respuesta**: El hiperparámetro que mayor impacto tuvo en el rendimiento del modelo fue `scale_pos_weight`, que tuvo una importancia de casi el $150\, \%$ más (es decir, casi la mitad más), que el primero que inmediatamente lo sigue en importancia: `learning_rate`. Por el gráfico de coordenadas paralelas, se pudo concluir que el mejor _recall_ se alcanzó con valores de `scale_pos_weight` $\sim 3\text{,}2$. Este hiperparámetro, en esencia, regula el peso de la clase positiva en el problema de optimización. Que sea $>1$ quiere decir que se está asignando un peso mucho más grande a equivocarse al clasificar las muestras realmente positivas (_falsos negativos_), entonces, es consistente con el desbalance de clases.

- ¿Cuánto mejoró el rendimiento del modelo después de la optimización de hiperparámetros?

A continuación, se usará Optuna para una grilla de hiperparámetros que busca optimizar _recall_, bajo el modelo `LightGBM` seleccionado en el ítem anterior. Esta grilla considerará parámetros del modelo y del _column transformer_ asociado al preprocesamiento.

```python
import optuna
from sklearn.metrics import recall_score
from sklearn.preprocessing import StandardScaler

def objective_function(trial, X_train, y_train, X_val, y_val, preprocessor, SEED=42) -> float:
  """
  Objective function for Optuna hyperparameter optimization.

  Parameters
  ----------
  trial : optuna.trial.Trial
    An Optuna trial object.
  X_train : pd.DataFrame
    Training features.
  y_train : pd.Series
    Training labels.
  X_val : pd.DataFrame
    Validation features.
  y_val : pd.Series
    Validation labels.
  preprocessor : ColumnTransformer
    Preprocessing pipeline.
  SEED : int, optional
    Random seed for reproducibility, by default 42.

  Returns
  -------
  float
    The macro recall score on the validation set.
  """
  params = {
    "clf__num_leaves": trial.suggest_int("num_leaves", 20, 200),
    "clf__max_depth": trial.suggest_int("max_depth", 3, 12),
    "clf__learning_rate": trial.suggest_float("learning_rate", 0.001, 0.1, log=True),
    "clf__n_estimators": trial.suggest_int("n_estimators", 50, 500),
    "clf__min_data_in_leaf": trial.suggest_int("min_data_in_leaf", 10, 100),
    "clf__feature_fraction": trial.suggest_float("feature_fraction", 0.5, 1.0),
    "clf__bagging_fraction": trial.suggest_float("bagging_fraction", 0.5, 1.0),
    "clf__bagging_freq": trial.suggest_int("bagging_freq", 1, 10),
    "clf__lambda_l1": trial.suggest_float("lambda_l1", 0.0, 10.0),
    "clf__lambda_l2": trial.suggest_float("lambda_l2", 0.0, 10.0),
    "clf__scale_pos_weight": trial.suggest_float("scale_pos_weight", 0.5, 5.0),
    "clf__subsample": trial.suggest_float("subsample", 0.5, 1.0),
    "clf__colsample_bytree": trial.suggest_float("colsample_bytree", 0.5, 1.0),
  }

  params.update({
    "preprocess__num__imputer__strategy": trial.suggest_categorical(
      "num_imputer_strategy", ["mean", "median"]
    ),
  })

  # Tries different choices for numerical scaler.
  scaler_choice = trial.suggest_categorical("num_scaler", ["minmax", "standard", "none"])
  if scaler_choice == "minmax":
    params["preprocess__num__scaler"] = MinMaxScaler()
  elif scaler_choice == "standard":
    params["preprocess__num__scaler"] = StandardScaler()
  else:
    params["preprocess__num__scaler"] = "passthrough"

  # Tries different encoders for cat_low features.
  cat_low_choice = trial.suggest_categorical("cat_low_encoder", ["onehot", "target"])
  if cat_low_choice == "onehot":
    # Other hyperparameters of OneHotEncoder are not relevant for optimization.
    encoder = OneHotEncoder(handle_unknown="ignore")
  else:
    encoder = ce.TargetEncoder(
      smoothing=trial.suggest_float("cat_low_target_smoothing", 1.0, 20.0),
      min_samples_leaf=trial.suggest_int("cat_low_target_min_leaf", 1, 20)
    )
  params["preprocess__cat_low__encoder"] = encoder

  params.update({
    "preprocess__cat_high__target__smoothing": trial.suggest_float("cat_high_smoothing", 1.0, 20.0),
    "preprocess__cat_high__target__min_samples_leaf": trial.suggest_int("cat_high_min_samples_leaf", 1, 20),
  })

  model = Pipeline(steps=[
    ('preprocess', preprocessor),
    ('clf', LGBMClassifier(objective='binary', random_state=SEED))
  ])
  model.set_params(**params)

  model.fit(X_train, y_train)

  y_pred = model.predict(X_val)
  recall = recall_score(y_val, y_pred, average='macro')

  return recall
```

Se usarán 20 _trials_ dada la cantidad de combinaciones que hay.
